## Demo 04: Using Python to Interact with Data in Files
It is a common requirement to interact with data located on various storage devices (e.g., local hard drives, corporate network-available storage devices & storage area networks (SAN's), cloud-based storage, (DataLakes)). File-based data may be stored in a variety of formats including: plain text, tab-separated values, comma-separated values, JSON (JavaScript Object Notation), XML (eXtensible Markup Language), as well as binary document filetypes, audio-visual filetypes. This demonstration illustrates methods for interacting with some of these data formats as they are written to, and read from, various text-based files.

Since we must leverage the underlying Operating System to gain access to any storage devices, local or network-available, we will first **import** the appropriate **os** library. 

```python
import os
```

### 1.0. Interacting with Text Files
#### 1.1. Writing a List to a Text File

```python
# Create a list to hold some agenda items.
agenda = ["Wake Up at 6:30 AM",
          "Stretch and exercise",
          "Take a shower",
          "Eat breakfast",
          "Read about Python programming"
         ]

# Specify a location for creating the new text file.
data_dir = os.path.join(os.getcwd(), 'data')
agenda_file = os.path.join(data_dir, 'agenda_entries.txt')

# Open the new file, in write mode, to capture the new data.
file = open(agenda_file, "w") # 'w'=write mode, 'r'=read mode, 'a'=append mode

# Write the contents of the list to the file.
file.write("\n".join(agenda))

# The file must be 'closed' to save the appended data.
file.close()
```

#### 1.2. Reading Data from the Newly Created Text File

```python
# Open the new file, in read mode, to access the data.
file = open(agenda_file, "r")  # 'r'=read mode
content = file.read()
file.close()

print(content)
```

#### 1.3. Appending Data to an Existing Text File

```python
# Create some additional agenda items
new_agenda_items = ["Commute to Work",
                    "Get Coffee with a Coworker",
                    "Start a New Python Project",
                    "Do Some Python Research"
                   ]

# Specify the location of the existing text file.
agenda_file = os.path.join(data_dir, 'agenda_entries.txt')

# Open the existing file, in append mode, to capture the new items.
file = open(agenda_file, 'a')  # 'a'=append mode

# Write the contents of the list to the file.
file.write("\n" + "\n".join(new_agenda_items))

# The file must be 'closed' to save the appended data.
file.close()
```

#### 1.4. Read New Data from the Text File

```python
# Open the new file, in read mode, to access the data.
file = open(agenda_file, "r")  # 'r'=read mode
content = file.read()
file.close()

# Verify the additional agenda items in the file
print(content)
```

#### 1.5. Find Entries Containing 'Python' and Write them to a New Text File

```python
# Open the new file, in read mode, to access the data.
with open(agenda_file, "r") as file:
    entries = file.read().split("\n")

python_entries = []
for entry in entries:
    if "Python" in entry:
        python_entries.append(entry)

# Specify a location for creating the new text file.
python_agenda_file = os.path.join(data_dir, 'python_agenda_entries.txt')

# Open the new file, in write mode, to capture the new data.
with open(python_agenda_file, "w") as file:
    # Write the contents of the list to the file.
    file.write("\n".join(python_entries))

# Open the new file, in read mode, to verify the data.
with open(python_agenda_file, "r") as file:
    print(file.read())
```

### 2.0. Writing a List of Dictionaries to a New File
The previous demonstration illustrated writing and reading very simple data to a plain text file; however, it is far more common to interact with data stored in one of the semantically self-describing data formats like comma-separated values (CSV), or JavaScript Object Notation (JSON). Building on the previous example, this demonstration will illustrate how to interact with these storage formats using the appropriate data transformation techniques and file types (MIME's).

#### 2.1. Create a List of Dictionaries to Hold Some Agenda Items
Dictionaries are a very commonly implemented data structure. They store data in a semantically self-describing format, and they provide the basis for the JavaScript Object Notation (JSON) format. Multi-line JSON files store multiple dictionaries contained within a List, and are a very common data structure for persisting tabular data to data files. 

```python
agenda = [
    {
        "agenda_id" : 1,
        "activity" : "Wake Up",
        "time" : "6:30 AM"
    },
    {
        "agenda_id" : 2,
        "activity" : "Stretch and exercise",
        "time" : "6:35 AM"
    },
    {
        "agenda_id" : 3,
        "activity" : "Take a shower",
        "time" : "6:50 AM"
    },
    {
        "agenda_id" : 4,
        "activity" : "Eat breakfast",
        "time" : "7:15 AM"
    },
    {
        "agenda_id" : 5,
        "activity" : "Read about Python programming",
        "time" : "8:00 AM"
    }
]
```

#### 2.2. Convert the List of Dictionaries to Comma-Separated Values and Write Them to a CSV File
Historically, another very popular data interchange format is comma-separated values (CSV) because it is leveraged by so many applications used to input, store and represent tabular data. It is commonly used by spreadsheet applications like Microsoft Excel, but is even more broadly adopted as a data storage and interchange format, and is used extensively with data extract-transform-load (ETL) pipeline processes.

```python
# Specify a location for creating the new text file.
data_dir = os.path.join(os.getcwd(), 'data')
agenda_csv_file = os.path.join(data_dir, 'agenda_entries.csv')

# Open the new file, in write mode, to capture the new data.
file = open(agenda_csv_file, "w") # 'w'=write mode, 'r'=read mode, 'a'=append mode

# Write the contents of the list of dictionaries to the file in CSV format with headers.
file.write("\n".join([",".join([f"{value}" for value in d.values()]) for d in [dict(zip(agenda[0], agenda[0].keys()))] + agenda]))

# The file must be 'closed' to save the appended data.
file.close()
```

#### 2.3. Reading from the Newly Created CSV File

```python
# Open the new file, in read mode, to access the data.
file = open(agenda_csv_file, "r")  # 'r'=read mode
content = file.read()
file.close()

print(content)
```

This validates that the data extracted from the list of dictioneries was successfully translated from its native JavaScript Object Notation to comma-separated values. Either of these are actually interchange formats that preserve the semantic context along with the data.

#### 2.4. Appending Data to an Existing CSV File

```python
# Create some additional agenda items
agenda = [
    {
        "agenda_id" : 6,
        "activity" : "Commute to Work",
        "time" : "8:30 AM"
    },
    {
        "agenda_id" : 7,
        "activity" : "Get Coffee with a Coworker",
        "time" : "9:00 AM"
    },
    {
        "agenda_id" : 8,
        "activity" : "Start a New Python Project",
        "time" : "9:15 AM"
    },
    {
        "agenda_id" : 9,
        "activity" : "Do Some Python Research",
        "time" : "9:45 AM"
    }
]

# Specify the location of the existing CSV file.
agenda_csv_file = os.path.join(data_dir, 'agenda_entries.csv')

# Open the existing file, in append mode, to capture the new items.
file = open(agenda_csv_file, 'a')  # 'a'=append mode

# Write the contents of the list of dictionaries to the file in CSV format without headers.
file.write("\n" + "\n".join([','.join([f"{v}" for v in d.values()]) for d in agenda]))

# The file must be 'closed' to save the appended data.
file.close()
```

#### 2.5. Read Newly Added Data From the CSV File

```python
# Open the new file, in read mode, to access the data.
file = open(agenda_csv_file, "r")  # 'r'=read mode
content = file.read()
file.close()

print(content)
```

### 3.0. Working with JavaScript Object Notation (JSON) Data
In our previous example we illustrated how fundamental Python data constructs like the Dictionary and the List can be used to represent textual data in a semantically self-describing, semi-structured format that enables for nesting data structures and polyschematism. While we could certainly write the data to a file using native Python language elements, these tasks are more commonly accomplished using libraries like <b>json</b> and <b>csv</b>. The next demonstration will illustrate using the <b>Pandas</b> library to accomplish similar interchange tasks, but at a higher level of abstraction.
<p>To that end, first we must import the <b>json</b> and <b>csv</b> libraries.

```python
import json
import csv
```

#### 3.1. Extract Data from a CSV File

```python
# Specify the location of the existing CSV file.
agenda_csv_file = os.path.join(data_dir, 'agenda_entries.csv')

# Open the CSV file, read its contents, and convert its rows to dictionaries
with open(agenda_csv_file, mode='r', newline='', encoding='utf-8') as csvfile:
    data = list(csv.DictReader(csvfile))
```

#### 3.2. Write the Transformed CSV Data to a JSON File

```python
# Specify the location of the new JSON file
agenda_json_file = os.path.join(data_dir, 'agenda_entries.json')

# Serialize the data with indentation, and then write it to a JSON file.
with open(agenda_json_file, mode='w', encoding='utf-8') as jsonfile:
   json.dump(data, jsonfile, indent=4)
```

#### 3.3. Read the Data from the JSON File and Display Its File Formatting

```python
# Open the new file, in read mode, to access the data.
file = open(agenda_json_file, "r")  # 'r'=read mode
content = file.read()
file.close()

print(content)
```

### 4.0. Using Pandas to Interact with Data Files
The <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_csv.html#pandas.read_csv"><b>Pandas API</b></a> includes several functions that enable interacting with data in various file formats.  These include `read_csv()` & `to_csv()`, `read_excel()` & `to_excel()`, `read_html()` & `to_html()`, `read_json()` & `to_json()`, and `read_xml()` & `to_xml()`... to name a few. What's more, this library provides critical functionality that enables textual data to be manipulated as a tabular object... similar to that of a spreadsheet application.

The following demonstration will illustrate the pattern common to file interaction with these functions using the comma-separated values (CSV) file format as an example.

#### 4.1. Load Data from a CSV File

```python
import pandas as pd

data_dir = os.path.join(os.getcwd(), 'data')
data_file = os.path.join(data_dir, 'northwind_products.csv')

df = pd.read_csv(data_file, header=0, index_col=0)
df.head()
```

#### 2.2. Drop Unimportant Features

```python
df.drop(['description','attachments'], axis=1, inplace=True)
df.columns
```

#### 2.3. Add a New Feature Using a Map (Dictionary) to Assign New Values

```python
df['discontinued_desc'] = df.discontinued.map({1: 'True', 0: 'False'})
df.head()
```

#### 2.4. Identify Duplicated Observations

```python
df[df.duplicated(subset='product_code', keep=False) == True]
```

#### 2.5. Write the Contents of the DataFrame to a New JavaScript Object Notation (JSON) File

```python
new_file = os.path.join(data_dir, 'northwind_products_cleaned.json')
df.to_json(new_file)
```

#### 2.6. Open the New JavaScript Object Notation (JSON) File to Verify the Data

```python
df = pd.read_json(new_file)
df.head()
```
