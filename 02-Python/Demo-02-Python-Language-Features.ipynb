## Demo 2: Python Language Features
Python has numerous liguistic devices to help you produce more elegant and more highly performant code.

#### 1.0. The 'Walrus' Operator
This special operator **`( := )`** lets you assign and use a value by way of the same expression

```python
# The old, more verbose, way
line = file.readline()
while line:
    process(line)
    line = file.readline()

# The Walrus operator cleans this up:
while line := file.readline():
    process(line)
```

#### 2.0. The 'In' Test
Rather than stacking multiple **OR** conditions, the **in** clause enables you to replace more verbose checking constructs

```python
# The common approach
status = "new"
if status == "new" or status == "pending" or status == "completed":
    print(status)

# The more elegant approach
status = "completed"
if status in {"new", "pending", "completed"}:
    print(status)
```

#### 3.0. The Dict.get() function
Rather than constructing repeated "check if exists" conditionals, use the **`.get()`** method to extract one of many members from a dictionary

```python
# The common approach
if "theme" in config:
    theme = config["theme"]
else:
    theme = "light"

# The more elegant approach
theme = config.get("theme", "light")
```

#### 4.0. List Comprehensions
Rather than using manual looping constructs, a list comprehension will be more declaritive and perhaps even faster

```python
# The common approach
squares = []
for x in values:
    squares.append(x * x)

# The more elegant approach
squares = [x * x for x in values]
```

#### 5.0. Enumerators
Rather than using manual index counters, an enumerator grants access to the index and value in a more concise manner 

```python
items = ["first item", "second item"]

# The common approach
i = 0
for item in items:
    print(i, item)
    i += 1

# The more elegant approach
for i, item in enumerate(items):
    print(i, item)
```

#### 6.0. Multiple Return Values (Tuple unpacking)
Rather than returning a dictionary, or creating temporary variables, unpack multiple returns to keep code tight

```python
# The common approach
res = get_user()
name = res[0]
age = res[1]

# The more elegant approach
name, age = get_user()
```

#### 7.0. The **`_`** Operator
You can use the `_` operator allows you to skip unwanted elements

```python
The _, status, message = api_call()

# OR
for _, value in items:
    print(value)
```

#### 8.0. The **`any()`** and **`all()`** Functions
Rather than constructing multiple nested conditions, these functions convert complex logic into more expressive statements

```python
a = 1
b = 2
c = 3

# The common approach
if (a > 0 and b > 0 and c > 0):
    print(f"Is 'a' greater than 'b'? {a > b}.")

# The more elegant approach
if all(x > 0 for x in (a, b, c)):
    print(f"Is 'a' greater than 'b'? {a > b}.")
```

#### 9.0. Swap variables without using a temp variable

```python
# The common approach
x = 1
y = 2

temp = x
x = y
y = temp

print(f"The value of 'x' is: {x}, and the value of 'y' is: {y}.")


# The more elegant approach
x = 1
y = 2

x, y = y, x

print(f"The value of 'x' is: {x}, and the value of 'y' is: {y}.")
```

#### 10.0. **`f-strings`**
Rather than authoring complex concatenations, f-strings will clean up your code. They're especially useful when authoring debugging expressions.

```python
name = "Robert"
age = 48

# The common approach
print("User: " + name + " (" + str(age) + ")")

# The more elegant approach
print(f"User: {name} ({age})")

# Or maybe even better
print(f"{name=}, {age=}")
```

#### 11.0. **`pathLib`**
Rather than using `os.path`, **`pathLib`** will produce cleaner and more readable code. It's also chainable and object-oriented

```python
from pathlib import Path

data = Path("data.json")
if data.exists():
    content = data.read_text()
```

## 12.0. Decorators
<p>As part of the <a href="https://www.freecodecamp.org/news/solid-principles-explained-in-plain-english/"><b>SOLID</b></a> programming principles, object-oriented programming and design stipulates the <b>Single Responsibility Principle</b>, which states that any given function should implement only one single responsibility (i.e., element of functionality) that is semanticly consistant with the name of the function. Any other functionality, especially responsibilities that appear repeatedly in multiple functions, should be defined in dedicated functions and/or classes. This enforces the separation of responsibilities and increases the maintainability of the codebase. Any changes that may become necessary are centralized and are therefore inherited wherever the function is consumed.</p>
<p>Decorators are a Python language construct that enable adding common functionality to other functions by way of a simple declaration. This behavior is similar to what can be achieved by deriving from a base class that defines additonal functionality that should be common to all classes from which it is derived. The later approach describes how the <a href="https://en.wikipedia.org/wiki/Decorator_pattern">Decorator</a> design pattern works as per <i>(Design Patterns, Elements of Reusable Object-Oriented Software: Erich Gamma, Richard Helm, Ralph Johnson, John Vlissides).</i> However, in keeping with that volume's advice to prefer composition over inheritance, the declaritive approach enhances the flexibility of its application when compared to the 'lock-in' or 'brittleness' that results from adding the common functionality by way of inheritance.</p> 

### 12.1. Define the Decorator function

```python
# The outer function definition provides the 'handle' by which its added functionality may be accessed
# The outer function, which defines the decorator's interface, expects a reference to the calling (base) function
def timer_decorator(base_fn):
    
    # The core function defines how the added functionality is applied to the calling function
    # Passing the *args and **kwargs parameters ensure the decorator can 'pack' and handle both parameters and keyword parameters 
    def enhanced_fn(*args, **kwargs):
        # In this case, the added 'timer' functionality 'wraps' the call to the base function
        start_time = time.time()
        
        # The call to the base function 'unpacks' any args and/or keyword args
        # that may, or may not, have been passed. It also handles any return values that may,
        # or may not, be specified in the calling 'base' function.
        result = base_fn(*args, **kwargs)

        # The added 'timer' functionality 'wraps' the call to the base function
        end_time = time.time()

        print(f"Task time: {end_time - start_time} seconds")
        
        return result
    return enhanced_fn
```

### 12.2. Implement the Decorator
There are two methods for invoking a Decorator. The first is to make an explicit call to the Decorator's outer function, but this may be a bit cumbersome, and also negatively effects the 'readability' of your code.  The second approach involves simple `decorating` the consuming function with the name of the Decorator preceded by the `'@'` sign.

#### 12.2.1.1. Create the Functions without Decoration
In this example, functions are created without any `decorators`. The functionality defined in the `Decorator's` function ( <b>enhanced_fn</b> ) can only be accessed by explicitly calling the decorator's outer function ( <b>timer_decorator(base_fn)</b> ).

```python
import time
from datetime import datetime, timedelta
```

```python
def brew_coffee(coffee_type, brewing_time):
    print(f"Brewing {coffee_type} coffee...")
    time.sleep(brewing_time)
    print(f"{coffee_type} Coffee is ready!")

def make_coffee():
    print("Making Coffee...")
    time.sleep(1)
    print("Coffee is Ready to Drink!")
    return f"Return Value: Drink Coffee by {datetime.now() + timedelta(minutes=10)}"
```

#### 12.2.1.2. Explicitly Calling the Decorator's Outer Function
In these examples, a reference to the `Decorator` function is created first, and then the reference is called, passing in any parameters where expected.

```python
# First, create a reference to the Decorator function, passing a reference to the function to be modified.
brew_coffee = timer_decorator(brew_coffee)

# Then, call the Decorator function via the reference we created, passing the parameters it expects.
brew_coffee(coffee_type="French Roast", brewing_time=2)
```

```python
# First, create a reference to the Decorator function, passing a reference to the function to be modified.
make_coffee = timer_decorator(make_coffee)

# Then, call the Decorator function via the reference we created.
make_coffee()
```

#### 12.2.2.1. Create the Functions with Decoration
In this example, the functions are created and a reference to the `Decorator` function is made by way of a `decoration` (i.e., <b>@decorator_function</b>). This illustrates the prefered method of accessing `Decorator` functionality, where the additional responsibility is added simply by `decorating` the function definition with the name of the `Decorator's` outer function interface.

```python
@timer_decorator
def brew_coffee(coffee_type, brewing_time):
    print(f"Brewing {coffee_type} coffee...")
    time.sleep(brewing_time)
    print(f"{coffee_type} Coffee is ready!")

@timer_decorator
def make_coffee():
    print("Making Coffee...")
    time.sleep(1)
    print("Coffee is Ready to Drink!")
    return f"Return Value: Drink Coffee by {datetime.now() + timedelta(minutes=10)}"
```

#### 12.2.2.2. Calling a `Decorated` Function that Expects Parameters but Has No Return Value
In the following examples, the `Decorated` function is called by its interface; thereby, allowing access to the `decorator's` functionality without any explicit knowledge of how that functionality is implemented. Because the `decorated` function requires parameters, those values can be passed either positionally, or using the keyword parameter syntax. This enables a much cleaner function call that keeps code tidy and clearly understood.

##### 12.2.2.2.1. Passing Parameters handled by the `args` construct

```python
brew_coffee("French Roast", 2)
```

##### 12.2.2.2.2. Passing Key-Word Parameters handled by the `kwargs` construct

```python
brew_coffee(coffee_type="French Roast", brewing_time=2)
```

#### 12.2.2.3. Calling a `Decorated` Function Having a Return Value but Expecting No Parameters
In this example, the `decorated` function is called without requiring any knowledge of the functionality the `decorator` supplies. What's more, this function expects no parameters; however, the `Decorator` definition behaves as expected even though it contains logic to handle parameters.

```python
make_coffee()
```
